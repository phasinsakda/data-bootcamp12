##OOP

- attributes: name, species, age
- methods: function bark(), hbd()

```python
class Dog:
    def __init__(self, name, species, age):
        self.name = name
        self.species = species
        self.age = age
    def bark(self):
        print(f"{self.name}: Woof!")
    def hbd(self):
        self.age += 1
```

```python
## attributes
dog1 = Dog("milo","Chihuahua",3)
dog2 = Dog("david","Husky",6)
```

```python
print(f"{dog1.name} is {dog1.species}.")
print(f"{dog2.name} is {dog2.species}.")
dog1.bark()
```

```python
print(dog1.name+" is "+str(dog1.age)+" years old.")
dog1.hbd()
print(dog1.name+" is "+str(dog1.age)+" years old.")
```

```python
## Homework OOP for an ATM

## deposit, withdraw, check balance, OTP, order coffee, pay bill, internet etc.
```

```python
class ATM:
    def __init__(self, user_name, bank_name,balance):
        self.user_name = user_name
        self.bank_name = bank_name
        self.balance = balance
        print(f"{self.user_name} opens a new account at {self.bank_name}.")

    def __str__(self) -> str:
        return f"{self.user_name} use self.bank_name account and has {self.balance} THB."

    def check_balance(self):
        print(f"{self.user_name} has {self.balance} baht in their account.")

    def deposit(self,amount):
        self.balance += amount
        print(f"Deposit {amount} THB successfully!!"
              f"Your new balance is {self.balance} THB.")
```

```python
fc = ATM("Phasin","scb",1000)

print(fc)

fc.check_balance()

fc.deposit(500)
```

```python
## super class
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

## extension
class Employee(Person):
    def __init__(self, name, age, company):
        super().__init__(name,age)
        self.company = company
    def greeting(self):
        print("Hello world")
```

```python
person1 = Person("Phasin",28)
person1.greeting()
```

```python
person2 = Employee("Lisa",35,"google")
person2.greeting()
```

```python
# context manager เปิดปิดไฟล์อัตโนมัติ

read_data=[]

with open("hotel.csv","r") as file: ## r w a
    data = csv.reader(file)
    for row in data:
        read_data.append(row)
```

```python
# header
read_data[0]
```

```python
# Body data
read_data[1:]
```

```python
## for loop to calculate avg. price per night
prices = []

for row in read_data[1:]:
    prices.append(int(row[-1]))
print(f"Average price pernight: {sum(prices)/len(prices)}")
```

## Read&Write Files

```python
import csv

# write file + context manager
head = ["id", "name", "city"]
body = [
    [1, "John", "New York"],
    [2, "Jane", "London"],
    [3, "Jany", "London"],
]

with open("school.csv","w") as file:
    writer = csv.writer(file)
    writer.writerow(head)
    writer.writerows(body)
```

```python
## for loop
import requests
import time

characters=[]
for i in range(10):
    url = f"https://swapi.info/api/people/{i+1}"
    response = requests.get(url)
    response_json = response.json()
    data = [response_json["name"],
            response_json["height"],
            response_json["mass"],
            ]

    characters.append(data)
    #time.sleep(1)

print(characters)
```

```python
import pandas as pd

starwars = pd.DataFrame(characters, columns=["name","height","mass"])

starwars
```
